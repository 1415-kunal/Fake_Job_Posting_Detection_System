```python
import re
```

```python
def clean_text(text):
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Replace HTML line breaks and similar tags with a space
    text = re.sub(r"<br\s*/?>", " ", text)

    # Remove remaining HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Replace multiple whitespace characters with one space
    text = re.sub(r"\s+", " ", text)

    # Remove leading and trailing spaces
    text = text.strip()

    return text
```

```python
for col in text_columns:
    X[col] = X[col].apply(clean_text)
```

```python

duplicate_text_columns = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]

X["duplicate_group"] = (
    X[duplicate_text_columns]
    .fillna("")
    .astype(str)
    .agg(" || ".join, axis=1)
)

print("Total rows:", len(X))
print("Unique text groups:", X["duplicate_group"].nunique())
```

```python
print("\nGroups appearing more than once:")
print((group_sizes > 1).sum())
```

```python
group_label_counts = (
    pd.DataFrame({
        "duplicate_group": X["duplicate_group"],
        "target": y.values
    })
    .groupby("duplicate_group")["target"]
    .nunique()
)
```

```python
conflicting_groups = (group_label_counts > 1).sum()

print("Duplicate groups with conflicting labels:", conflicting_groups)
```

```python
print("\nTarget distribution:")
print(y.value_counts())

print("\nTarget percentage:")
print(y.value_counts(normalize=True).mul(100).round(2))
```

```python
from sklearn.model_selection import train_test_split

# Create one row per duplicate group
groups = X["duplicate_group"].unique()

print("Total unique duplicate groups:", len(groups))
```

```python
group_labels = (
    pd.DataFrame({
        "duplicate_group": X["duplicate_group"],
        "target": y.values
    })
    .groupby("duplicate_group")["target"]
    .first()
)

print(group_labels.value_counts())
```

```python
train_groups, test_groups = train_test_split(
    group_labels.index,
    test_size=0.20,
    random_state=42,
    stratify=group_labels.values
)

print("Training groups:", len(train_groups))
print("Testing groups:", len(test_groups))
```

```python
train_mask = X["duplicate_group"].isin(train_groups)
test_mask = X["duplicate_group"].isin(test_groups)

X_train = X.loc[train_mask].copy()
X_test = X.loc[test_mask].copy()

y_train = y.loc[train_mask].copy()
y_test = y.loc[test_mask].copy()

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
print("Total rows:", len(X_train) + len(X_test))
```

```python
train_group_set = set(X_train["duplicate_group"])
test_group_set = set(X_test["duplicate_group"])

overlap = train_group_set.intersection(test_group_set)

print("Duplicate groups appearing in both train and test:", len(overlap))
```
