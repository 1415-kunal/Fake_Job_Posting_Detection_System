```python
import re
```

```python
def clean_text(text):
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Replace HTML line breaks and similar tags with a space
    text = re.sub(r"<br\s*/?>", " ", text)

    # Remove remaining HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Replace multiple whitespace characters with one space
    text = re.sub(r"\s+", " ", text)

    # Remove leading and trailing spaces
    text = text.strip()

    return text
```

```python
for col in text_columns:
    X[col] = X[col].apply(clean_text)
```

```python
X["salary_available"] = (
    X["salary_range"] != "Unknown"
).astype(int)
```

```python
X["salary_min"] = pd.to_numeric(
    X["salary_range"].str.extract(r"(\d+)-(\d+)")[0],
    errors="coerce"
)
```

```python
X["salary_max"] = pd.to_numeric(
    X["salary_range"].str.extract(r"(\d+)-(\d+)")[1],
    errors="coerce"
)
```

```python
print(
    X[
        ["salary_range", "salary_available", "salary_min", "salary_max"]
    ].head(20)
)
```

```python
print("Salary ranges available:", X["salary_available"].sum())
print("Salary ranges missing:", (X["salary_available"] == 0).sum())
print("Salary min extracted:", X["salary_min"].notna().sum())
print("Salary max extracted:", X["salary_max"].notna().sum())
```

```python
# Extract country, region/state, and city from location

location_parts = X["location"].str.split(",")

X["location_country"] = location_parts.str[0].str.strip()
X["location_region"] = location_parts.str[1].str.strip()
X["location_city"] = location_parts.str[-1].str.strip()

# Handle empty values
for col in ["location_country", "location_region", "location_city"]:
    X[col] = X[col].replace("", "Unknown")
```

```python
X["location_unknown"] = (X["location"] == "Unknown").astype(int)
```
