```python
import re
```

```python
def clean_text(text):
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Replace HTML line breaks and similar tags with a space
    text = re.sub(r"<br\s*/?>", " ", text)

    # Remove remaining HTML tags
    text = re.sub(r"<[^>]+>", " ", text)

    # Replace multiple whitespace characters with one space
    text = re.sub(r"\s+", " ", text)

    # Remove leading and trailing spaces
    text = text.strip()

    return text
```

# Data Preprocessing Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

# Define the feature groups

```python
text_features = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]
```

```python
categorical_features = [
    "employment_type",
    "required_experience",
    "required_education",
    "industry",
    "function",
    "location_country",
    "location_region"
]
```

```python
numeric_features = [
    "telecommuting",
    "has_company_logo",
    "has_questions",
    "salary_available",
    "salary_min",
    "salary_max",
    "location_unknown",
    "department_available",
    "department_char_length",
    "department_word_count"
]
```

# combine the job text

```python
X_train["combined_text"] = (
    X_train[text_features]
    .fillna("")
    .agg(" ".join, axis=1)
)

X_test["combined_text"] = (
    X_test[text_features]
    .fillna("")
    .agg(" ".join, axis=1)
)
```

```python
print(X_train["combined_text"].head(1))
```

```python
X_train["combined_text"][20]
```

```python
tfidf = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    max_features=50000,
    sublinear_tf=True
)
```

```python
one_hot = OneHotEncoder(
    handle_unknown="ignore"
)
```

```python
numeric_transformer = SimpleImputer(
    strategy="median"
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "text",
            tfidf,
            "combined_text"
        ),
        (
            "categorical",
            one_hot,
            categorical_features
        ),
        (
            "numeric",
            numeric_transformer,
            numeric_features
        )
    ]
)
```

```python
print("Text feature:", "combined_text" in X_train.columns)

print("\nCategorical features:")
print(categorical_features)

print("\nNumeric features:")
print(numeric_features)
```

```python
X_train_transformed = preprocessor.fit_transform(X_train)
```

```python
X_test_transformed = preprocessor.transform(X_test)
```
